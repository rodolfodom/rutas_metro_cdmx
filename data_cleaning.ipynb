```python
import geopandas as gpd
import pandas as pd
import json
```

```python
geoshape = gpd.read_file('/Users/rodolfodom/Downloads/stcmetro_shp/STC_Metro_estaciones_utm14n.shp')
```

```python
geoshape.shape
```

```python
geoshape
```

```python
geoData = geoshape[['NOMBRE', 'LINEA', 'geometry']]
geoData
```

```python
duplicated_names = geoData[geoData['NOMBRE'].duplicated()]["NOMBRE"].unique()
duplicated_names
```

```python
geoData.loc[geoData['NOMBRE'].isin(duplicated_names), 'NOMBRE'] = geoData['NOMBRE'] + '-' + geoData['LINEA']
```

```python
geoData
```

```python
geoData["NOMBRE"] = geoData["NOMBRE"].str.replace("-0", "-")
geoData
```

```python
geoData["lat"] = geoData.geometry.y
geoData["lon"] = geoData.geometry.x
geoData
```

```python
geoData.rename(columns={'lon': 'lng'}, inplace=True)
```

```python
geoData.drop(columns=['geometry'], inplace=True, axis=1)
```

```python
geoData.to_csv('/Users/rodolfodom/rutas_metro_cdmx/coordinates.csv', index=False)
```

```python
geoData
```

```python
geodict = geoData[["NOMBRE", "lat", "lng"]].set_index('NOMBRE').T.to_dict('list')
geodict
```

```python
geodict_str = json.dumps(geodict, ensure_ascii=False)
with open('/Users/rodolfodom/rutas_metro_cdmx/coordinates_tuple.py', 'w', encoding='utf-8') as f:
    f.write(f'coordinates_tuple = {geodict_str}')
```
